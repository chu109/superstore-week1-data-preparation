```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("Superstore.csv")
df.head()
```

```python
print("Rows and columns:")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types and non-null counts:")
df.info()
```

```python
required_columns = [
    "Row ID",
    "Order ID",
    "Order Date",
    "Segment",
    "Region",
    "Category",
    "Sub-Category",
    "Product ID",
    "Product Name",
    "Sales",
    "Quantity",
    "Discount",
    "Profit"
]

df_analysis = df[required_columns].copy()

df_analysis.head()
```

```python
df_analysis["Order Date"] = pd.to_datetime(df_analysis["Order Date"])

df_analysis.info()
```

```python
start_date = pd.Timestamp("2026-01-01")
end_date = pd.Timestamp("2026-07-31")

date_check = df_analysis["Order Date"].between(start_date, end_date)

print("Minimum Order Date:", df_analysis["Order Date"].min())
print("Maximum Order Date:", df_analysis["Order Date"].max())
print("Rows outside Jan-Jul 2026:", (~date_check).sum())
```

```python
print("Missing values by column:")
print(df_analysis.isnull().sum())

print("\nDuplicated full rows:")
print(df_analysis.duplicated().sum())

print("\nDuplicated Row ID values:")
print(df_analysis["Row ID"].duplicated().sum())

print("\nUnusual values:")
print("Sales <= 0:", (df_analysis["Sales"] <= 0).sum())
print("Quantity <= 0:", (df_analysis["Quantity"] <= 0).sum())
print("Discount < 0:", (df_analysis["Discount"] < 0).sum())
print("Discount > 1:", (df_analysis["Discount"] > 1).sum())
print("Profit < 0:", (df_analysis["Profit"] < 0).sum())
```

```python
categorical_columns = ["Segment", "Region", "Category", "Sub-Category"]

for col in categorical_columns:
    print("\n" + col)
    print("Number of unique values:", df_analysis[col].nunique())
    print(df_analysis[col].value_counts())
```

```python
total_sales = df_analysis["Sales"].sum()
total_profit = df_analysis["Profit"].sum()

overall_profit_margin = total_profit / total_sales

loss_transaction_count = (df_analysis["Profit"] < 0).sum()
total_transactions = len(df_analysis)
loss_transaction_rate = loss_transaction_count / total_transactions

print("Total Sales:", round(total_sales, 2))
print("Total Profit:", round(total_profit, 2))
print("Overall Profit Margin:", round(overall_profit_margin * 100, 2), "%")
print("Loss Transactions:", loss_transaction_count)
print("Total Transactions:", total_transactions)
print("Loss Transaction Rate:", round(loss_transaction_rate * 100, 2), "%")
```

```python
df_analysis.to_csv("Superstore_analysis.csv", index=False, date_format="%Y-%m-%d")

print("Saved cleaned dataset as Superstore_analysis.csv")
print("Final dataset shape:", df_analysis.shape)
```
